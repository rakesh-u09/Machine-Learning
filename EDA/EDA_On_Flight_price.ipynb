FEATURES

The various features of the cleaned dataset are explained below:
1) Airline: The name of the airline company is stored in the airline column. It is a categorical feature having 6 different airlines.
2) Flight: Flight stores information regarding the plane's flight code. It is a categorical feature.
3) Source City: City from which the flight takes off. It is a categorical feature having 6 unique cities.
4) Departure Time: This is a derived categorical feature obtained created by grouping time periods into bins. It stores information about the departure time and have 6 unique time labels.
5) Stops: A categorical feature with 3 distinct values that stores the number of stops between the source and destination cities.
6) Arrival Time: This is a derived categorical feature created by grouping time intervals into bins. It has six distinct time labels and keeps information about the arrival time.
7) Destination City: City where the flight will land. It is a categorical feature having 6 unique cities.
8) Class: A categorical feature that contains information on seat class; it has two distinct values: Business and Economy.
9) Duration: A continuous feature that displays the overall amount of time it takes to travel between cities in hours.
10)Days Left: This is a derived characteristic that is calculated by subtracting the trip date by the booking date.
11) Price: Target variable stores information of the ticket price.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('economy.csv')

df.head()
```

```python
df.info()
```

Feature Engneering

```python
df['Date'] = df['date'].str.split('-').str[0]
df['Month'] = df['date'].str.split('-').str[1]
df['Year'] = df['date'].str.split('-').str[2]

df.head()
```

```python
df.info()
```

```python
df['Date'] = df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)

df.info()
```

```python
# Drop the original 'date' column as we have extracted the necessary information
df.drop('date', axis=1, inplace=True)
```

```python
df.head()
```

```python
df['Arrival_hour'] = df['arr_time'].str.split(':').str[0].astype(int)
df['Arrival_minute'] = df['arr_time'].str.split(':').str[1].astype(int)
df.head()
```

```python
# Drop the original 'arr_time' column as we have extracted the necessary information
df.drop('arr_time', axis=1, inplace=True)

df.head()
```

```python
# departure time feature engineering

df['Departure_hour'] = df['dep_time'].str.split(':').str[0].astype(int)
df['Departure_minute'] = df['dep_time'].str.split(':').str[1].astype(int)
df.info()
```

```python
# Drop the original 'dep_time' column as we have extracted the necessary information
df.drop('dep_time', axis=1, inplace=True)
df.head()
```

```python
# Time taken for the flight feature engineering

# Extract the numeric parts safely using regex
df['time_taken_hours'] = df['time_taken'].str.extract(r'(\d+)\s*h')
df['time_taken_minutes'] = df['time_taken'].str.extract(r'(\d+)\s*m')

# Convert using capital "Int64" which allows NaN/Missing values natively
df['time_taken_hours'] = df['time_taken_hours'].astype('Int64')
df['time_taken_minutes'] = df['time_taken_minutes'].astype('Int64')


df.head(10)
```

```python
df.info()
```

```python
# Drop the original 'time_taken' column as we have extracted the necessary information
df.drop('time_taken', axis=1, inplace=True)
```

```python
df.info()
```

```python
# convert the 'price' column to numeric type, handling any non-numeric values

df['price'] = df['price'].str.replace(',', '').astype(float)
```

```python
df.info()
```

```python
df.head()
```

```python
df['stop'] = df['stop'].str.strip()

df['stop'].unique()
```

```python
# Clean the stop value
def clean_stop(stop):
    if (stop == 'non-stop'):
        return '0'
    else:
        return stop[0]

df['stops'] = df['stop'].apply(clean_stop)

df.tail()
```

```python
df['stops'] = df['stops'].astype(int)
df.info()
```

```python
df.drop('stop', axis=1, inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
# Data Preprocessing 
df['airline'].unique()

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder()

encoded_df = pd.DataFrame(encoder.fit_transform(df[['airline', 'from', 'to']]).toarray(), columns=encoder.get_feature_names_out(['airline', 'from', 'to'])).head()

df = df.join(encoded_df)

df.head()
```

```python
df.info()
```

```python
# Remove the original categorical columns after encoding
df.drop(['airline', 'from', 'to'], axis=1, inplace=True)
```

```python
df.head()
```

```python
df.info()
```

```python
# Drop the ch_code column as it is not useful for prediction
df.drop('ch_code', axis=1, inplace=True)
```
